```python
import pickle
import numpy as np
from matplotlib import pyplot as plt
```

```python
def load_data(filename):
    with open(filename, 'rb') as f:
        data = pickle.load(f)
    return data

def plot_data(data, title):
    # plot 3 subfigures, one for both eps in columns 0 and 1, one for the D in columns 2 and 3 and the fitness vs the iterarions
    fig, axs = plt.subplots(3, 1, figsize=(10, 15))
    axs[0].plot(data["x"][:, 0], label='eps_p', color='tab:blue')
    axs[0].plot(data["x"][:, 1], label='eps_n', color='tab:orange')
    axs[0].axhline(eps_p_tar, color='tab:blue', linestyle='--', label='eps_p target')
    axs[0].axhline(eps_n_tar, color='tab:orange', linestyle='--', label='eps_n target')
    axs[0].set_xlabel('Iteration')
    axs[0].set_ylabel('Acive material volume fractions')
    axs[0].legend()
    axs[1].semilogy(data["x"][:, 2], label='D_p', color='tab:cyan')
    axs[1].axhline(D_p_tar, color='tab:cyan', linestyle='--', label='D_p target')
    axs[1].semilogy(data["x"][:, 3], label='D_n', color='tab:green')
    axs[1].axhline(D_n_tar, color='tab:green', linestyle='--', label='D_n target')
    axs[1].set_xlabel('Iteration')
    axs[1].set_ylabel('Diffusion Coefficient')
    axs[1].legend()
    axs[2].plot(data["fitness"], label='Fitness')
    axs[2].set_xlabel('Iteration')
    axs[2].set_ylabel('Fitness')
    axs[2].legend()
    plt.tight_layout()
```

```python
eps_p_tar = 0.7315000000000002 
eps_n_tar = 0.5625 
D_p_tar = 4e-13 
D_n_tar = 3.2999999999999997e-12
```

```python
# Load the data
data = load_data('optim_history.pkl')
data["GA"]["x"] = np.array(data["GA"]["x"])
data["GA"]["fitness"] = np.array(data["GA"]["fitness"])
# Plot the data
plot_data(data["GA"], 'Optimization GAPSO Evaluation Results')
```

```python
data = load_data('optim_history_pso.pkl')
data["PSO"]["x"] = np.array(data["PSO"]["x"])
data["PSO"]["fitness"] = np.array(data["PSO"]["fitness"])
# Plot the data
plot_data(data["PSO"], 'Optimization GAPSO Evaluation Results')
```
